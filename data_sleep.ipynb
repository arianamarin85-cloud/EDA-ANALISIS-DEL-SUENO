# Análisis de los factores de salud y estilo de vida que impactan en la calidad y duración del sueño.

## 1. Introducción y Marco Teórico

El descanso es uno de los pilares básicos del bienestar diario, junto con una alimentación equilibrada y la actividad física. Según informes de instituciones como la Sociedad Española de Neurología (SEN), hasta un 45% de la población experimenta problemas relacionados con el sueño en algún momento de su vida.

Dormir mal o descansar menos de las 7 a 9 horas diarias recomendadas en adultos suele reflejarse rápidamente en el día a día: menor concentración, mayor sensación de cansancio y un estado de ánimo más irritable. 

El sueño cumple funciones vitales como consolidar la memoria y regular las hormonas, la presión arterial, las defensas y las emociones, por lo que descansar mal aumenta el riesgo de sufrir obesidad, diabetes, hipertensión, problemas vasculares y cáncer (Fernández Arcos, s. f.). Asimismo, dormir pocas horas deteriora la atención y la agilidad mental a corto plazo, mientras que a largo plazo eleva el riesgo de desarrollar trastornos neurológicos —como cefaleas, ictus, Parkinson o Alzheimer— y enfermedades mentales como ansiedad, depresión o psicosis (Fernández Arcos, s. f.).

Dado que el sueño no depende de un único factor, sino de la combinación de hábitos y estilo de vida, este estudio se enfoca en tres áreas cotidianas:

1. **El estrés:** Las tensiones del día a día y la carga mental suelen ser el principal enemigo para conciliar un sueño profundo y reparador.
2. **Las diferencias entre hombres y mujeres:** Diversos estudios señalan que los patrones de descanso, las rutinas diarias y las cargas emocionales no afectan por igual a ambos géneros.
3. **El peso y la condición física:** Mantener un peso saludable y realizar actividad física regular se asocia comúnmente con un descanso más continuo y de mejor calidad, mientras que el sobrepeso suele vincularse a dificultades para dormir bien.

## Exploración del contenido de las variables a utilizar.

```python
# Revisamos si hay valores duplicados.
duplicates = df[df.duplicated()]
print(duplicates)
```

```python
# Verificamos si hay valores nulos.
df.isnull().any()
```

```python
# Obtenemos la cantidad de valores NAN de la columna "Sleep Disorder" = 219 rows 
df[df["Sleep Disorder"].isnull()]
```

```python
# Renombramos las filas de la columna "Sleep Disorder" de valores "NAN" por No_Disorder" 
df_limpio = df.fillna("No_Disorder")
df_limpio
```

```python
# La columna "Blood Pressure", la separamos en "SYS,"DIA" para manipular de forma más limpia nuestros datos
df_limpio[["SYS", "DIA"]] = (df_limpio["Blood Pressure"].str.split("/", n=1, expand=True).astype(int))
df_limpio
```

```python
#Eliminamos la columna "Blood Pressure"

df_limpio = df_limpio.drop("Blood Pressure", axis = 1)
df_limpio
```

```python
print("Genero: ", df_limpio["Gender"].unique())
print("Duración del Sueño: ", df_limpio["Sleep Duration"].unique())
print("Calidad del Sueño: ", df_limpio["Quality of Sleep"].unique())
print("IMC: ", df_limpio["BMI Category"].unique())
```

```python
# Detectamos y corregimos la duplicidad de "Normal" vs "Normal Weight"
df_limpio = df_limpio.replace(to_replace = ["Normal"], value = ["Normal Weight"])
print("IMC: ", df_limpio["BMI Category"].unique())
```
