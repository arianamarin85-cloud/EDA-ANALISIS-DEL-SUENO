### 6.1. Origen del Conjunto de Datos
Los datos utilizados para este proyecto provienen de la plataforma **Kaggle**, correspondientes al conjunto de datos titulado *"Sleep Health and Lifestyle Dataset"*. Este conjunto recopila métricas de hábitos de vida, indicadores de salud y calidad de descanso de una muestra poblacional diversa.

| Variable | Descripción |
| --- | --- |
| Person ID | Un identificador para cada individuo. |
| Gender | El género de la persona (Masculino/Femenino). |
| Age | La edad de la persona en años. |
| Occupation | La ocupación o profesión de la persona. |
| Sleep Duration (hours) | El número de horas que la persona duerme por día. |
| Sleep Quality (scale: 1-10) | Una calificación subjetiva de la calidad del sueño, que va de 1 a 10. |
| Physical Activity Level (minutes/day) | El número de minutos que la persona realiza actividad física diariamente. |
| Stress Level (scale: 1-10) | Una calificación subjetiva del nivel de estrés experimentado por la persona, que va de 1 a 10. |
| BMI Category | La categoría de IMC de la persona (p. ej., Bajo peso, Normal, Sobrepeso). |
| Blood Pressure (systolic/diastolic) | La medición de la presión arterial de la persona, indicada como presión sistólica sobre presión diastólica. |
| Heart Rate (bpm) | La frecuencia cardíaca en reposo de la persona en latidos por minuto. |
| Daily Steps | El número de pasos que la persona da por día. |
| Sleep Disorder | Presencia o ausencia de un trastorno del sueño en la persona (Ninguno, Insomnio, Apnea del sueño). |

## Exploración del contenido de las variables a utilizar.

### Exploración de Valores Duplicados



```python
# Revisamos si hay valores duplicados, excluyendo la columna "Person ID" para identificar si hay perfiles repetidos con distintos IDs
mascara = df.drop(columns=["Person ID"]).duplicated()
df[mascara]
```

```python
# Detectamos y corregimos la duplicidad de "Normal" vs "Normal Weight"
df_limpio = df.copy()
df_limpio["BMI Category"] = df_limpio["BMI Category"].replace({"Normal": "Normal Weight"})
df_limpio["BMI Category"].value_counts()
```

### Resultados de la exploración de Valores Duplicados 
    Del contenido total del dataset que son 374 registros, encontramos la cantidad de 242 registros duplicados (al excluir la columna Person ID), por lo que nos quedarían únicamente un total de 132 registros únicos. Dado que este es un dataset sintético podría considerarse que el creador del mismo ha utilizado registros idénticos para representar un patrón frecuente. No haremos limpieza de estos datos tomando la consideración anterior pero tomaremos en cuenta que afectarán al estudio de la correlación ya que estarán las variables más relacionadas que en un estudio real.
    
    * Variable BMI Category:
    Normalizamos las variables con elementos duplicados en este caso "Normal" y "Normal Weight" que son valores idénticos.
    Identificamos 3 tipos de registros para esta categoria donde tendremos especial cuidado al analizar la categoria "Obese", debido a la poca cantidad de registros de la misma.
    

### Exploración de Valores Nulos

```python
# Verificamos si hay valores nulos.
df.isnull().any()
```

```python
# Obtenemos la cantidad de valores NAN de la columna "Sleep Disorder" = 219 rows 
df[df["Sleep Disorder"].isnull()]
```

```python
# Renombramos las filas de la columna "Sleep Disorder" de valores "NAN" por No_Disorder" 
df_limpio["Sleep Disorder"] = df_limpio["Sleep Disorder"].fillna("No_Disorder")
df_limpio.isnull().sum()
```

## Revisión de valores Nulos:
  
La columna con registros incompletos es "Sleep Disorder", que contiene solo **155 valores no nulos**. Los 219 registros restantes corresponden a personas que no presentan diagnóstico de Trastorno de sueño, por lo que se ha procedido a imputar los valores nulos por el valor de "No_Disorder".

Según la información de datos recopilados de **Kaggle**, se indican detalles sobre la columna de trastornos del sueño "Sleep Disorder": 

  - Ninguno(NAN): El individuo no presenta ningún trastorno del sueño específico.

  - Insomnio: La persona experimenta dificultad para conciliar el sueño o para permanecer dormida, lo que conlleva un sueño insuficiente o de mala calidad.

  - Apnea del sueño: La persona sufre pausas en la respiración durante el sueño, lo que provoca alteraciones en los patrones de sueño y posibles riesgos para la salud.

```python
# La columna "Blood Pressure", la separamos en "SYS,"DIA" para manipular de forma más limpia nuestros datos
df_limpio[["SYS", "DIA"]] = (df_limpio["Blood Pressure"].str.split("/", n=1, expand=True).astype(int))
df_limpio
```

```python
# revisamos si los registros de la tensión corresponden, siendo la SYS menor o igual que DIA
df_limpio[df_limpio["SYS"] <= df_limpio["DIA"]]
```

```python
#Eliminamos la columna "Blood Pressure"
df_limpio = df_limpio.drop("Blood Pressure", axis = 1)
df_limpio
```
