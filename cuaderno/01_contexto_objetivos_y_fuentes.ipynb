# Impacto de los factores de salud y estilo de vida en la calidad y duración del sueño.

## 1. Introducción

El descanso es uno de los pilares básicos del bienestar diario, junto con una alimentación equilibrada y la actividad física. Según informes de instituciones como la Sociedad Española de Neurología (SEN), hasta un 45% de la población experimenta problemas relacionados con el sueño en algún momento de su vida.

Dormir mal o descansar menos de las 7 a 9 horas diarias recomendadas en adultos suele reflejarse rápidamente en el día a día: menor concentración, mayor sensación de cansancio y un estado de ánimo más irritable. 

El sueño cumple funciones vitales como consolidar la memoria y regular las hormonas, la presión arterial, las defensas y las emociones, por lo que descansar mal aumenta el riesgo de sufrir obesidad, diabetes, hipertensión, problemas vasculares y cáncer (Fernández Arcos, s. f.). Asimismo, dormir pocas horas deteriora la atención y la agilidad mental a corto plazo, mientras que a largo plazo eleva el riesgo de desarrollar trastornos neurológicos —como cefaleas, ictus, Parkinson o Alzheimer— y enfermedades mentales como ansiedad, depresión o psicosis (Fernández Arcos, s. f.).

Dado que el sueño no depende de un único factor, sino de la combinación de hábitos y estilo de vida, este estudio se enfoca en las siguientes áreas de la vida cotidiana:

1. **El estrés:** Las tensiones del día a día y la carga mental suelen ser el principal enemigo para conciliar un sueño profundo y reparador.
2. **Las diferencias entre hombres y mujeres:** Diversos estudios señalan que los patrones de descanso, las rutinas diarias y las cargas emocionales no afectan por igual a ambos géneros.
3. **El peso y la condición física:** Mantener un peso saludable y realizar actividad física regular se asocia comúnmente con un descanso más continuo y de mejor calidad, mientras que el sobrepeso suele vincularse a dificultades para dormir bien.

## 3. Objetivos del Proyecto

### Objetivo General
Analizar cómo influyen factores cotidianos (el nivel de estrés, el peso corporal, género, edad, la ocupación profesional y los trastornos del sueño) en las horas y la calidad de sueño de un grupo de personas, utilizando técnicas de análisis exploratorio de datos (EDA).

Los datos que presentados son sintéticos y han sido creados con fines ilustrativos.

### Objetivos Específicos
1. **Evaluar el impacto del estrés:** Comprobar si las personas que reportan niveles altos de estrés duermen con menor calidad.
2. **Comparar por género:** Identificar si existe una diferencia real en la cantidad de horas que duermen hombres y mujeres.
3. **Analizar el peso corporal:** Determinar si las personas con sobrepeso u obesidad presentan una peor calidad de sueño frente a quienes tienen un peso normal.
4. **Conocer la relación IMC-Trastorno de Sueño**Analizar la asociacion bidireccional que existe entre el IMC y los trastornos del sueño para comprobar si ambas condiciones se relacionan proporcionalmente entre sí.
5. **Conocer la influencia de la Ocupación**Comparar los niveles de estrés, la duración y calidad del sueño entre las distintas ocupaciones profesionales, para identificar cuales son los perfiles más afectados. 
6. **Examinar la relación de la Edad****Determinar si los grupos más jovenes presentan mayores niveles de estrés y una menor duración y calidad de descanso en comparación con los rangos de edades más altas.
