# 9. Conclusiones Generales y Recomendaciones

## 9.1. Hallazgos Principales por Hipótesis

1. **El estrés es el factor más determinante en la calidad del sueño (Hipótesis 1 : Confirmada):**

   Existe una relación negativa (-0.90) entre el estrés y la calidad del sueño, a medida que aumenta el estrés, disminuye la calidad del sueño. Quienes mantienen un menor nivel de estrés(entre 3 y 5 puntos) logran una calidad del sueño mucho mayor, con valores promedio de entre 7.67 y 8.97, mientras que en el nivel máximo de estrés (8 puntos) la calidad del sueño cae hasta un 5.86.
   
2. **El género no marca una diferencia relevante en las horas de descanso (Hipótesis 2 : Descartada en la práctica):**

   Aunque las mujeres presentan una media ligeramente superior a la de los hombres (7.23 h frente a 7.04 h), apenas 11 minutos de diferencia) y una mayor dispersión = 0.88 vs. 0.69), ambos grupos comparten exactamente la misma mediana (7.20 h) y sus distribuciones se solapan ampliamente.

3. **El exceso de peso deteriora la calidad del descanso (Hipótesis 3: Confirmada):**

   La calidad media del sueño desciende conforme aumenta el índice de masa corporal, 7.64 en personas con peso normal, 6.90 en sobrepeso y 6.40 en obesidad.
   
4. **Fuerte asociación bidireccional entre IMC y trastornos del sueño (Hipótesis 4: Confirmada):**

   * *De trastorno a peso:* Aproximadamente 9 de cada 10 personas diagnosticadas con insomnio (88.3%) o apnea del sueño (91.0%) tienen sobrepeso u obesidad, frente a solo el 8.7% en quienes no presentan trastornos.

   * *De peso a trastorno:* El 87.1% de las personas con sobrepeso y el 100% de los casos registrados con obesidad padecen algún trastorno del sueño, mientras que el 92.6% de las personas con peso normal duermen sin patologías diagnosticadas. 
   
5. **La ocupación influye en el estrés-duración de sueño (Hipótesis 5: Confirmada):**

   Las ocupaciones como: *Doctor* (estrés medio 6.73; sueño 6.97 h) y *Salesperson* (estrés 7.00; sueño 6.40 h) se ven más afectadas , frente a perfiles como *Engineer* (estrés 3.89; sueño 7.99 h) o *Accountant* (estrés 4.59; sueño 7.11 h).

6. **Los adultos jóvenes soportan mayor estrés y peor descanso (Hipótesis 6 : Confirmada):**
   El tramo de **27 a 35 años** registra mayor nivel estrés medio (6.57), menor calidad (6.50) y menor duración de sueño (6.85 h). Conforme avanza la edad hacia el grupo de **46+ años**, el estrés cae a 4.75 y el descanso mejora hasta las 7.50 horas y una calidad media de 7.96.


---
## 9.2. Conclusiones 

Tras analizar las 374 registros y comprobar las 6 hipótesis del estudio, se extraen las siguientes conclusiones:

La salud del sueño no depende del género, sino de una combinación de factores vinculados al estrés, el trabajo, la edad y el peso corporal.
En particular, el estrés es el factor que más perjudica el descanso, afectando especialmente a los adultos jóvenes y a profesiones de mayor presión (como el ámbito sanitario y comercial). En cambio, en edades más avanzadas y en ocupaciones más estables, el estrés disminuye y el sueño mejora de forma clara. A esto se suma el exceso de peso, que también se relaciona estrechamente con un peor descanso y con una mayor presencia de trastornos del sueño.

En definitiva, dormir mal no es una cuestión de género, sino principalmente de estrés, carga laboral y sobrepeso; factores sobre los que sí es posible actuar mediante medidas dirigidas a los grupos más vulnerables.

## 9.3. Limitaciones del Estudio

* **Naturaleza sintética y registros repetidos:** 

Al excluir "Personal Id", el dataset cuenta con 132 perfiles únicos y 242 filas repetidas que representan patrones frecuentes, lo que tiende a intensificar las correlaciones frente a una muestra poblacional real.

* **Falta de balance para grupos específicos:** 

Las conclusiones sobre la categoría "Obese" (10 registros) y profesiones como "Manager" (1 registros), "Sales Representative" (2 registros), "Scientist" (4 registros) y "Software Engineer" (4 registros) debemos analizarlas con mayor cautela por su bajo tamaño muestral.

## 9.4. Recomendaciones y Próximos Pasos

- Priorizar programas de higiene del sueño y gestión del estrés dirigidos al segmento adulto joven (27–35 años) y a colectivos laborales de alto estrés (personal sanitario y equipos comerciales).
- Integrar el seguimiento del peso corporal en protocolos de detección temprana de apnea del sueño e insomnio.
