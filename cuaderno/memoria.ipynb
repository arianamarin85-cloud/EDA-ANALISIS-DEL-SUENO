# Impacto de los factores de salud y estilo de vida en la calidad y duración del sueño.

## 1. Introducción 

El descanso es uno de los pilares básicos del bienestar diario, junto con una alimentación equilibrada y la actividad física. Según informes de instituciones como la Sociedad Española de Neurología (SEN), hasta un 45% de la población experimenta problemas relacionados con el sueño en algún momento de su vida.

Dormir mal o descansar menos de las 7 a 9 horas diarias recomendadas en adultos suele reflejarse rápidamente en el día a día: menor concentración, mayor sensación de cansancio y un estado de ánimo más irritable. 

El sueño cumple funciones vitales como consolidar la memoria y regular las hormonas, la presión arterial, las defensas y las emociones, por lo que descansar mal aumenta el riesgo de sufrir obesidad, diabetes, hipertensión, problemas vasculares y cáncer (Fernández Arcos, s. f.). Asimismo, dormir pocas horas deteriora la atención y la agilidad mental a corto plazo, mientras que a largo plazo eleva el riesgo de desarrollar trastornos neurológicos —como cefaleas, ictus, Parkinson o Alzheimer— y enfermedades mentales como ansiedad, depresión o psicosis (Fernández Arcos, s. f.).

Dado que el sueño no depende de un único factor, sino de la combinación de hábitos y estilo de vida, este estudio se enfoca en las siguientes áreas de la vida cotidiana:

1. **El estrés:** Las tensiones del día a día y la carga mental suelen ser el principal enemigo para conciliar un sueño profundo y reparador.
2. **Las diferencias entre hombres y mujeres:** Diversos estudios señalan que los patrones de descanso, las rutinas diarias y las cargas emocionales no afectan por igual a ambos géneros.
3. **El peso y la condición física:** Mantener un peso saludable y realizar actividad física regular se asocia comúnmente con un descanso más continuo y de mejor calidad, mientras que el sobrepeso suele vincularse a dificultades para dormir bien.

## 8.4. Prueba de Hipótesis 4 
 *"Las personas con un trastorno del sueño diagnosticado presentan una mayor proporción de sobrepeso u obesidad; y, a su vez, las personas con sobrepeso u obesidad presentan una mayor proporción de trastornos del sueño."*
* **Variables a considerar:** "BMI Category", "Sleep Disorder".

```python
#creamos una tabla y agregamos la funcion aggfunc="size" que nos cuente cuantas filas hay en cada cruce de los datos fill_value=0 si no existe alguna combinación
df_h4 = df_limpio[["BMI Category", "Sleep Disorder"]].copy()
resultado_h4 = df_h4.pivot_table(index = "BMI Category", columns = "Sleep Disorder", aggfunc="size", fill_value=0)
resultado_h4
```

```python
#sabemos el IMC calculamos, calculamos el % trastorno
total_filas = resultado_h4.sum(axis=1)  # Suma horizontal: total de personas en cada peso
pct_filas = resultado_h4.div(total_filas, axis=0) * 100
pct_filas.round(1)
```

```python
#sabemos el trastorno, calculamos el % sobre su IMC  
total_col = resultado_h4.sum(axis=0)  # # Suma vertical: total de personas en cada trastorno
pct_col = resultado_h4.div(total_col, axis=1) * 100
pct_col.round(1)
```

```python
# visualizamos la comparación de cuantas personas tienen cada trastorno del sueño según su categoría de IMC
plt.figure(figsize=(6, 4))
sns.countplot(data=df_h4, x="Sleep Disorder", hue="BMI Category")
plt.title("Trastornos del sueño según categoría de IMC")
plt.savefig("../imagen/trastornos_sueno_imc.jpg")
plt.show()
```

```python
#compruebo los resultados obtenidos en el gráfico
df_h4.value_counts()
```

### Podemos obtener la relación entre las variables:
Según los datos registrados y mediante el cruce de las variables, determinamos que existe una relación bidireccional entre ambas, aunque no podemos concluir cuál ha provocado a cuál.

Se concluye que:
* De trastorno a peso: 9 de cada 10 personas con un trastorno diagnosticado tienen sobrepeso u obesidad (el "88.3%" en insomnio y el "91.0%" en apnea del sueño), frente a solo el "8.7%" en las personas sin trastorno.
* De peso a trastorno: Entre las personas con "sobrepeso", el "87.1%" presenta insomnio o apnea, y en "obesidad" sube al "100%" (teniendo en cuenta que la muestra de obesidad es de solo 10 registros). En cambio, en las personas con "peso normal" solo el "7.4%" tiene algún trastorno (el "92.6%" no presenta ninguno).
En el gráfico de barras ("Trastornos del sueño según categoría de IMC") se observa que el grupo sin trastornos ("No_Disorder") está formado casi por completo por personas con peso normal ("200" frente a solo "19" con sobrepeso y "0" con obesidad), mientras que en "Insomnio" y  Apnea" ocurre lo contrario y predomina el sobrepeso ("64" y "65" personas frente a "9" y "7" con peso normal). Además, aunque las barras de obesidad se ven bajas porque solo hay "10" registros en el estudio, todos ellos ("4" con insomnio y "6" con apnea) presentan algún trastorno del sueño.

## 9.1. Hallazgos Principales por Hipótesis

1. **El estrés es el factor más determinante en la calidad del sueño (Hipótesis 1 : Confirmada):**

   Existe una relación negativa (-0.90) entre el estrés y la calidad del sueño, a medida que aumenta el estrés, disminuye la calidad del sueño. Quienes mantienen un menor nivel de estrés(entre 3 y 5 puntos) logran una calidad del sueño mucho mayor, con valores promedio de entre 7.67 y 8.97, mientras que en el nivel máximo de estrés (8 puntos) la calidad del sueño cae hasta un 5.86.
   
2. **El género no marca una diferencia relevante en las horas de descanso (Hipótesis 2 : Descartada en la práctica):**

   Aunque las mujeres presentan una media ligeramente superior a la de los hombres (7.23 h frente a 7.04 h), apenas 11 minutos de diferencia) y una mayor dispersión = 0.88 vs. 0.69), ambos grupos comparten exactamente la misma mediana (7.20 h) y sus distribuciones se solapan ampliamente.

3. **El exceso de peso deteriora la calidad del descanso (Hipótesis 3: Confirmada):**

   La calidad media del sueño desciende conforme aumenta el índice de masa corporal, 7.64 en personas con peso normal, 6.90 en sobrepeso y 6.40 en obesidad.
   
4. **Fuerte asociación bidireccional entre IMC y trastornos del sueño (Hipótesis 4: Confirmada):**

   * *De trastorno a peso:* Aproximadamente 9 de cada 10 personas diagnosticadas con insomnio (88.3%) o apnea del sueño (91.0%) tienen sobrepeso u obesidad, frente a solo el 8.7% en quienes no presentan trastornos.

   * *De peso a trastorno:* El 87.1% de las personas con sobrepeso y el 100% de los casos registrados con obesidad padecen algún trastorno del sueño, mientras que el 92.6% de las personas con peso normal duermen sin patologías diagnosticadas. 
   
5. **La ocupación influye en el estrés-duración de sueño (Hipótesis 5: Confirmada):**

   Las ocupaciones como: *Doctor* (estrés medio 6.73; sueño 6.97 h) y *Salesperson* (estrés 7.00; sueño 6.40 h) se ven más afectadas , frente a perfiles como *Engineer* (estrés 3.89; sueño 7.99 h) o *Accountant* (estrés 4.59; sueño 7.11 h).

6. **Los adultos jóvenes soportan mayor estrés y peor descanso (Hipótesis 6 : Confirmada):**
   El tramo de **27 a 35 años** registra mayor nivel estrés medio (6.57), menor calidad (6.50) y menor duración de sueño (6.85 h). Conforme avanza la edad hacia el grupo de **46+ años**, el estrés cae a 4.75 y el descanso mejora hasta las 7.50 horas y una calidad media de 7.96.
